## Tratamento de erros

```python
try:
    result = 10 / 0
except ZeroDivisionError:
    print("Cannot divide by zero!")
```

```python
try:
    file = open("./file.txt", "r")
    print(file.read())
except FileNotFoundError:
    print("File not found!")
finally:
    file.close()
```

## Arquivos

```python
file = open("./file.txt", "w")
file.write("Hello, World!")
file.close()
file = open("./file.txt", "r")
print(file.read())
file.close()
```
