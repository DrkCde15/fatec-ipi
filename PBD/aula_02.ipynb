# LIST COMPRERENTION

```python
numeros=[1,2,3,4,5,6]
quadrados = [x**2 for x in numeros if x % 1 == 0]
print(quadrados)
```

# SET

```python
conjunto1 = {1, 2, 3}
conjunto2 = {3, 4, 5}

uniao = conjunto1 | conjunto2 # | é o operador de união
print(uniao)
```

```python
intersecao = conjunto1 & conjunto2 # & é o operador de interseção
print(intersecao)
```

```python
diferenca = conjunto1 - conjunto2 # - é o operador de diferença
print(diferenca)
```

```python
diferenca_simetrica = conjunto1 ^ conjunto2 # ^ é o operador de diferença simétrica
print(diferenca_simetrica)
```

```python
frutas = {"maçã", "banana", "laranja"}
frutas.add("pera") # add é o método para adicionar um elemento a um conjunto
print(frutas)
```

```python
frutas.remove("banana") # remove é o método para remover um elemento de um conjunto. Se o elemento não existir, ele gera um erro.
print(frutas)
```

```python
frutas.discard("uva")# discard é o método para remover um elemento de um conjunto. Se o elemento não existir, ele não gera um erro.
print(frutas)
```

```python
frutas.clear() # clear é o método para remover todos os elementos de um conjunto
print(frutas)
```

# FUNCOES

```python
def saudacao(nome):
    print(f"Olá, {nome}, seja bem-vindo!")
saudacao("Julio")
```

```python
def soma(a, b):
    return a + b
resultado = soma(3, 5)
print(resultado)
```

```python
def subtracao(a, b):
    return a - b
resultado = subtracao(5, 3)
print(resultado)
```

```python
def multiplicacao(a, b):
    return a * b
resultado = multiplicacao(3, 5)
print(resultado)
```

```python
def divisao(a, b):
    return a / b
resultado = divisao(10, 2)
print(resultado)
```
